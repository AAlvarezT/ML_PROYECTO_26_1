**Proyecto:** Encuesta Permanente de Empleo Nacional (EPEN)  
**Objetivo:** Cargar y documentar el dataset original sin ninguna modificación, estableciendo la fuente de verdad del proyecto.

## 1. Descripción de la fuente de datos

- **Fuente:** Encuesta Permanente de Empleo Nacional (EPEN)
- **Institución:** Instituto Nacional de Estadística y Geografía (INEGI) / Organismo equivalente
- **Período de referencia:** Indicar trimestre/año correspondiente
- **Unidad de observación:** Persona en edad de trabajar (14 años o más)
- **Cobertura geográfica:** Nacional

### Variables disponibles (resumen)
| Variable | Descripción | Tipo |
|---|---|---|
| `edad` | Edad en años cumplidos | Numérica |
| `sexo` | Sexo del entrevistado | Categórica |
| `nivel_educativo` | Último nivel de estudios aprobado | Categórica |
| `estado_civil` | Situación conyugal | Categórica |
| `ingreso_mensual` | Ingreso laboral mensual (MXN) | Numérica |
| `horas_trabajadas` | Horas trabajadas a la semana | Numérica |
| `tipo_empleo` | Formal / Informal | Categórica |
| `sector` | Sector económico | Categórica |
| `condicion_actividad` | Ocupado / Desocupado / No PEA | Categórica (target) |

```python
# ─── Librerías ────────────────────────────────────────────────────────────────
import pandas as pd
import numpy as np
import os
```

```python
# ─── Rutas ────────────────────────────────────────────────────────────────────
# Ajustar la ruta al archivo original según el entorno de trabajo
DATA_PATH = os.path.join('..', 'data', 'raw', 'epen_dataset.csv')

# ─── Carga del dataset ────────────────────────────────────────────────────────
# df = pd.read_csv(DATA_PATH, encoding='utf-8')
# Descomenta la línea anterior cuando el archivo esté disponible.

# Dataset de ejemplo para validar el flujo del notebook
np.random.seed(42)
n = 1000
df = pd.DataFrame({
    'edad': np.random.randint(14, 70, n),
    'sexo': np.random.choice(['Hombre', 'Mujer'], n),
    'nivel_educativo': np.random.choice(
        ['Sin instrucción', 'Primaria', 'Secundaria', 'Preparatoria', 'Universidad', 'Posgrado'], n
    ),
    'estado_civil': np.random.choice(['Soltero', 'Casado', 'Unión libre', 'Divorciado', 'Viudo'], n),
    'ingreso_mensual': np.random.exponential(8000, n).round(2),
    'horas_trabajadas': np.random.randint(0, 60, n),
    'tipo_empleo': np.random.choice(['Formal', 'Informal', np.nan], n, p=[0.45, 0.45, 0.10]),
    'sector': np.random.choice(
        ['Agricultura', 'Industria', 'Comercio', 'Servicios', 'Gobierno'], n
    ),
    'condicion_actividad': np.random.choice(
        ['Ocupado', 'Desocupado', 'No PEA'], n, p=[0.60, 0.10, 0.30]
    ),
})

print(f'Dataset cargado: {df.shape[0]:,} filas × {df.shape[1]} columnas')
df.head()
```
