**Proyecto:** Encuesta Permanente de Empleo Nacional (EPEN)  
**Objetivo:** Revisar y documentar el significado, tipo, rango y calidad de cada variable del dataset.

```python
import pandas as pd
import numpy as np
import os
import sys
sys.path.insert(0, os.path.join('..', '..'))

# Cargar datos (snapshot del raw data)
RAW_PATH = os.path.join('..', 'data', 'raw', 'epen_snapshot.csv')
try:
    df = pd.read_csv(RAW_PATH)
except FileNotFoundError:
    # Datos de ejemplo si aún no se generó el snapshot
    np.random.seed(42)
    n = 1000
    df = pd.DataFrame({
        'edad': np.random.randint(14, 70, n),
        'sexo': np.random.choice(['Hombre', 'Mujer'], n),
        'nivel_educativo': np.random.choice(
            ['Sin instrucción', 'Primaria', 'Secundaria', 'Preparatoria', 'Universidad', 'Posgrado'], n
        ),
        'estado_civil': np.random.choice(['Soltero', 'Casado', 'Unión libre', 'Divorciado', 'Viudo'], n),
        'ingreso_mensual': np.random.exponential(8000, n).round(2),
        'horas_trabajadas': np.random.randint(0, 60, n),
        'tipo_empleo': np.random.choice(['Formal', 'Informal', np.nan], n, p=[0.45, 0.45, 0.10]),
        'sector': np.random.choice(['Agricultura', 'Industria', 'Comercio', 'Servicios', 'Gobierno'], n),
        'condicion_actividad': np.random.choice(['Ocupado', 'Desocupado', 'No PEA'], n, p=[0.60, 0.10, 0.30]),
    })
print(f'Filas: {df.shape[0]:,} | Columnas: {df.shape[1]}')
```

## 1. Diccionario de Variables

```python
diccionario = pd.DataFrame([
    {'Variable': 'edad',               'Tipo': 'Numérica continua',  'Descripción': 'Edad en años cumplidos al momento de la encuesta', 'Rango/Valores': '14 – 99'},
    {'Variable': 'sexo',               'Tipo': 'Categórica nominal', 'Descripción': 'Sexo biológico del entrevistado', 'Rango/Valores': 'Hombre, Mujer'},
    {'Variable': 'nivel_educativo',    'Tipo': 'Categórica ordinal', 'Descripción': 'Último nivel de estudios aprobado', 'Rango/Valores': 'Sin instrucción → Posgrado'},
    {'Variable': 'estado_civil',       'Tipo': 'Categórica nominal', 'Descripción': 'Situación conyugal del entrevistado', 'Rango/Valores': 'Soltero, Casado, Unión libre, Divorciado, Viudo'},
    {'Variable': 'ingreso_mensual',    'Tipo': 'Numérica continua',  'Descripción': 'Ingreso laboral mensual en pesos mexicanos (MXN)', 'Rango/Valores': '0 – sin límite superior'},
    {'Variable': 'horas_trabajadas',   'Tipo': 'Numérica discreta',  'Descripción': 'Número de horas trabajadas en la semana de referencia', 'Rango/Valores': '0 – 168'},
    {'Variable': 'tipo_empleo',        'Tipo': 'Categórica nominal', 'Descripción': 'Formalidad del empleo principal', 'Rango/Valores': 'Formal, Informal'},
    {'Variable': 'sector',             'Tipo': 'Categórica nominal', 'Descripción': 'Sector económico donde labora', 'Rango/Valores': 'Agricultura, Industria, Comercio, Servicios, Gobierno'},
    {'Variable': 'condicion_actividad','Tipo': 'Categórica nominal', 'Descripción': 'Condición de actividad económica (variable objetivo)', 'Rango/Valores': 'Ocupado, Desocupado, No PEA'},
])
diccionario
```

## 2. Calidad de datos básica

```python
calidad = pd.DataFrame({
    'Variable': df.columns,
    'Tipo_pandas': df.dtypes.values,
    'Nulos': df.isnull().sum().values,
    'Nulos_%': (df.isnull().mean() * 100).round(2).values,
    'Únicos': df.nunique().values,
})
calidad
```
