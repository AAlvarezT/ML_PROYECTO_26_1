# 03 – Preprocessing: División Train/Test

**Proyecto:** Encuesta Permanente de Empleo Nacional (EPEN)  
**Objetivo:** Dividir el dataset en conjuntos de entrenamiento y prueba, respetando la distribución del target (stratified split) para garantizar representatividad de ambas clases.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import os

# ─── Cargar datos codificados ──────────────────────────────────────────────────
ENC_PATH = os.path.join('..', 'data', 'processed', 'epen_encoded.csv')
try:
    df = pd.read_csv(ENC_PATH)
except FileNotFoundError:
    np.random.seed(42)
    n = 1000
    df = pd.DataFrame(
        np.random.randn(n, 10),
        columns=[f'feature_{i}' for i in range(10)]
    )
    df['target_desocupado'] = np.random.choice([0, 1], n, p=[0.90, 0.10])

TARGET = 'target_desocupado'
FEATURES = [c for c in df.columns if c != TARGET]

X = df[FEATURES]
y = df[TARGET]

print(f'Observaciones totales: {len(X):,}')
print(f'Distribución del target:\n{y.value_counts(normalize=True).round(3)}')
```

## 1. División estratificada 80/20

```python
RANDOM_STATE = 42
TEST_SIZE = 0.20

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y,
)

print(f'Entrenamiento : {X_train.shape[0]:,} filas ({X_train.shape[0]/len(X)*100:.1f}%)')
print(f'Prueba        : {X_test.shape[0]:,} filas ({X_test.shape[0]/len(X)*100:.1f}%)')
print(f'\nDistribución target – Train:\n{y_train.value_counts(normalize=True).round(3)}')
print(f'Distribución target – Test :\n{y_test.value_counts(normalize=True).round(3)}')
```

## 2. Guardar los conjuntos

```python
SPLIT_DIR = os.path.join('..', 'data', 'split')
os.makedirs(SPLIT_DIR, exist_ok=True)

X_train.to_csv(os.path.join(SPLIT_DIR, 'X_train.csv'), index=False)
X_test.to_csv(os.path.join(SPLIT_DIR, 'X_test.csv'), index=False)
y_train.to_csv(os.path.join(SPLIT_DIR, 'y_train.csv'), index=False)
y_test.to_csv(os.path.join(SPLIT_DIR, 'y_test.csv'), index=False)

print('Archivos guardados en:', SPLIT_DIR)
print('  X_train.csv, X_test.csv, y_train.csv, y_test.csv')
```
